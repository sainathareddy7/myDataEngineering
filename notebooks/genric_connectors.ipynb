```python
def spark_connector():   
    spark.conf.set(
    "fs.azure.account.key.optumadlssg.dfs.core.windows.net",
    dbutils.secrets.get(scope = 'optumscope', key = 'adlskey'))
    return "Connected to adls"

 
```

```python
def read_bronze_csv_data(fn):
    df = spark.read.csv("abfss://optum@optumadlssg.dfs.core.windows.net/bronze/"+fn+".csv",inferSchema=True,header=True)
    return df
```

```python
def read_silver_csv_data(fn):
    df = spark.read.csv("abfss://optum@optumadlssg.dfs.core.windows.net/silver/"+fn+".csv",inferSchema=True,header=True)
    return df
```

```python
def read_bronze_json_data(fn):
    df = spark.read.json("abfss://optum@optumadlssg.dfs.core.windows.net/bronze/"+fn+".json")
    return df
```

```python
def write2silver(df,file_name):
    #file_name="Hospital_s.csv"
    silver_path = "abfss://optum@optumadlssg.dfs.core.windows.net/silver/"
    temp_path = "abfss://optum@optumadlssg.dfs.core.windows.net/silver/output_temp"
    final_path=f"{silver_path}{file_name}"
    display(df)
    df.coalesce(1).write.mode("overwrite").csv(temp_path,header=True)
    files= dbutils.fs.ls(temp_path)
    csv_files=[file.path for file in files if file.name.endswith(".csv")]
    dbutils.fs.mv(csv_files[0],final_path)
    dbutils.fs.rm(temp_path,recurse=True)
    print("CSV File written sucessfully to silver path")
```
