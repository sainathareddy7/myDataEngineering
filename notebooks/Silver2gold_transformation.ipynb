```python
from pyspark.sql.functions import col

hos_data = hos_data.withColumnRenamed("city", "h_city")
pat_data = pat_data.withColumnRenamed("city", "p_city")

# Use DataFrame aliases and dot notation for columns
pat_hos_df = pat_data.alias("pat").join(
    hos_data.alias("hosp"),
    col("pat.hospital_id") == col("hosp.Hospital_id"),
    "inner"
)

pat_hos_df = pat_hos_df.drop("Hospital_id")
pat_hos_df = pat_hos_df.withColumnRenamed("state", "h_state") \
                       .withColumnRenamed("country", "h_country")

display(pat_hos_df)
```
